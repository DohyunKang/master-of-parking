```python
from ultralytics import YOLO
from IPython.display import display, Image
import os

# YOLO 모델 불러오기
model = YOLO('/content/runs/detect/train/weights/best.pt')

def predict_and_display(image_path):
    # 예측 수행 및 결과 저장
    results = model.predict(source=image_path, save=True, project='/content/runs', name='predict_results')


# 예측할 이미지 경로 설정
image_paths = ['/content/IMG_0588.JPG','/content/IMG_0600.JPG','/content/IMG_0408.JPG','/content/IMG_0425.JPG']  # 각 이미지 파일 경로를 여기에 추가

# 각 이미지에 대해 예측 수행 및 결과 표시
for image_path in image_paths:
    predict_and_display(image_path)
```
